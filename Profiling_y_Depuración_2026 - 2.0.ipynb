verificaciones antes del merge

```python
# ==========================================
# NORMALIZAR NOMBRES DE COMUNAS
# ==========================================

def normalizar_comuna(valor):
    if pd.isna(valor):
        return pd.NA

    valor = str(valor).strip().lower()

    # Quitar acentos
    valor = unicodedata.normalize("NFD", valor)
    valor = "".join(
        c for c in valor
        if unicodedata.category(c) != "Mn"
    )

    # Reemplazar caracteres especiales por espacios
    valor = re.sub(r"[^a-z0-9]+", " ", valor)

    # Eliminar espacios múltiples
    valor = re.sub(r"\s+", " ", valor).strip()

    return valor


# ==========================================
# CREAR CLAVES NORMALIZADAS
# ==========================================

df["_comuna_normalizada"] = df["ComunaUnidad"].apply(
    normalizar_comuna
)

comunas_regiones["_comuna_normalizada"] = (
    comunas_regiones["nombre_comuna"]
    .apply(normalizar_comuna)
)


# ==========================================
# CORREGIR ALIAS DE COMUNAS
# ==========================================

alias_comunas = {
    "santiago centro": "santiago",
    "la calera": "calera",
    "puerto saavedra": "saavedra",
    "puerto natales": "natales",
    "san vicente de tagua tagua": "san vicente",
    "til til": "tiltil",
    "alto bio bio": "alto biobio",
    "chol chol": "cholchol",
    "torres del payne": "torres del paine",
    "las guaitecas": "guaitecas"
}

df["_comuna_normalizada"] = (
    df["_comuna_normalizada"]
    .replace(alias_comunas)
)
```

Realizar Merge Comuna a Región

```python
df = df.merge(
    comunas_regiones[
        ["_comuna_normalizada", "nombre_region"]
    ],
    how="left",
    on="_comuna_normalizada"
)

df.rename(
    columns={"nombre_region": "Region"},
    inplace=True
)

df.loc[
    df["ComunaUnidad"] == "Trehuaco",
    "Region"
] = "Región de Ñuble"
```

```python
df["Valor Total Ofertado"] = np.where(
    df["Moneda Adquisicion"] == "Peso Chileno",
    df["Valor Total Ofertado"],

    np.where(
        df["Moneda Adquisicion"] == "Unidad de Fomento",
        df["Valor Total Ofertado"] * df["UF_CLP"],

        np.where(
            df["Moneda Adquisicion"] == "Dolar",
            df["Valor Total Ofertado"] * df["USD_CLP"],

            np.nan
        )
    )
)
```
